```python
import sys
from pathlib import Path

# Ensure project root is importable
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data.loader import load_reviews
from src.preprocessing.cleaner import preprocess_dataframe
from src.sentiment.analyzer import run_sentiment_pipeline
from src.aspects.extractor import AspectExtractor, get_aspect_sentiment_summary
from src.topics.modeler import run_topic_pipeline
from src.keywords.extractor import run_keyword_pipeline
from src.clustering.embedder import run_clustering_pipeline
from src.search.semantic_search import SemanticSearch
from src.insights.intelligence import generate_insights
```

## 1. Load Data

```python
sample_csv = ROOT / 'data' / 'raw' / 'sample_reviews.csv'
df = load_reviews(str(sample_csv))
print(f'Loaded {len(df)} reviews')
df.head()
```

## 2. Text Preprocessing

```python
df = preprocess_dataframe(df)
df[['original_review', 'cleaned_review', 'model_input']].head()
```

## 3. Sentiment Analysis

```python
df = run_sentiment_pipeline(df, use_transformer=False)
print(df['sentiment'].value_counts())
df[['original_review', 'vader_sentiment', 'vader_compound', 'uncertainty_flag']].head()
```

## 4. Aspect-Based Sentiment Analysis

```python
extractor = AspectExtractor()
aspects_df = extractor.extract_from_dataframe(df)
df = extractor.add_aspects_summary(df, aspects_df)
summary = get_aspect_sentiment_summary(aspects_df)
summary.head(10)
```

## 5. Topic Discovery & Keywords

```python
df, topic_summary, _ = run_topic_pipeline(df, prefer_bertopic=False)
keywords = run_keyword_pipeline(df, use_keybert=False)
display(topic_summary[['topic_label', 'count', 'keywords']].head())
display(keywords.head(10))
```

## 6. Clustering & Semantic Search

```python
df, embeddings, method = run_clustering_pipeline(df)
searcher = SemanticSearch(embeddings, df, method)
results = searcher.search('battery issues and fast charging', top_k=3)
results[['original_review', 'similarity_score', 'sentiment', 'aspects_summary']]
```
